```python
%matplotlib inline
import numpy as np
import galsim

import jax
jax.config.update("jax_enable_x64", True)
import jax.numpy as jnp
import matplotlib.pylab as plt
```

```python
ngrid = 50
g1 = 0.0; g2 = 0.00
obj = galsim.Sersic(n=1.3, half_light_radius=1.5, flux =20.0).shear(g1=g1, g2=g2)
psf = galsim.Gaussian(sigma=2.0); obj = galsim.Convolve(obj, psf)
im = obj.shift(0.5,0.5).drawImage(nx=ngrid, ny=ngrid, scale=1, method = 'no_pixel').array.astype(np.float64)

plt.close()
fig, ax= plt.subplots(1, 1, figsize=(4, 4))
ax.imshow(im, cmap='RdYlBu_r')
ax.set_xticklabels([]); ax.set_yticklabels([])
print("Simulate Galaxy Image:")
plt.show()

ngrid = 50
g1 = 0.01; g2 = 0.00
obj = galsim.Sersic(n=1.3, half_light_radius=1.5, flux =20.0).shear(g1=0,g2=0).shear(g1=g1, g2=g2)
psf = galsim.Gaussian(sigma=2.0); obj = galsim.Convolve(obj, psf)
im_shear = obj.shift(0.5,0.5).drawImage(nx=ngrid, ny=ngrid, scale=1, method = 'no_pixel').array.astype(np.float64)

plt.close()
fig, ax= plt.subplots(1, 1, figsize=(4, 4))
ax.imshow(im_shear, cmap='RdYlBu_r')
ax.set_xticklabels([]); ax.set_yticklabels([])
print("Sheared Galaxy Image:")
plt.show()

plt.close()
fig, ax= plt.subplots(1, 1, figsize=(4, 4))
ax.imshow(im_shear-im, cmap='RdYlBu_r')
ax.set_xticklabels([]); ax.set_yticklabels([])
print("Diff Galaxy Image:")
plt.show()
```

```python
def get_pixel_response(data, sigma):
    ny, nx = data.shape
    k2grid, k1grid = jnp.meshgrid(
        jnp.fft.fftfreq(ny, 1 / np.pi / 2.0), 
        jnp.fft.fftfreq(nx, 1 / np.pi / 2.0), 
        indexing="ij",
    )
    # convolution
    imF = jnp.fft.fft2(data)
    # shear response of smoothed image
    imFQ1 = imF*(k1grid**2.-k2grid**2.) * sigma**2.
    imFD1 = imF*(-1j*k1grid)
    imFD2 = imF*(-1j*k2grid)
    # back to configuration space
    imCovQ1 = jnp.fft.ifft2(imFQ1).real
    imCovD1 = jnp.fft.ifft2(imFD1).real
    imCovD2 = jnp.fft.ifft2(imFD2).real
    # Create 2D grid values using meshgrid
    ygrid, xgrid = jnp.meshgrid(
        jnp.arange(-ny/2, ny/2), 
        jnp.arange(-nx/2, nx/2),
        indexing='ij',
    )
    return imCovQ1+xgrid*imCovD1-ygrid*imCovD2
pixel_response = get_pixel_response(im, sigma=2.0)

plt.close()
fig, ax= plt.subplots(1, 1, figsize=(4, 4))
ax.imshow(pixel_response, cmap='RdYlBu_r')
ax.set_xticklabels([]); ax.set_yticklabels([])
print("The pixel shear response dpix / dg_1 is:")
plt.show()

plt.close()
fig, ax= plt.subplots(1, 1, figsize=(4, 4))
ax.imshow(pixel_response*0.01-(im_shear-im), cmap='RdYlBu_r')
ax.set_xticklabels([]); ax.set_yticklabels([])
print("The residual is:")
#plt.colorbar()
plt.show()
```
